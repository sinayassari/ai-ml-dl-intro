##Resahping, Viewing, Stacking, Squeezing, Unsqueezing

```python
#Create a tensor
import torch
tensor_C = torch.arange(1, 13)
tensor_C, tensor_C.shape
```

```python
tensor_C_reshaped = tensor_C.reshape(1, 9)
tensor_C_reshaped, tensor_C_reshaped.shape
```

```python
tensor_C_reshaped = tensor_C.reshape(3, 4)
tensor_C_reshaped, tensor_C_reshaped.shape
```

```python
x_stacked = torch.stack([tensor_C, tensor_C, tensor_C, tensor_C], dim=0)
x_stacked
```

```python
x_stacked = torch.stack([tensor_C, tensor_C, tensor_C, tensor_C], dim=1)
x_stacked
```

```python
tensor_C_Squeezed = tensor_C.squeeze(dim=0)
tensor_C, tensor_C_Squeezed, tensor_C_Squeezed.shape
```

```python
tensor_C_Unsqueezed = tensor_C_Squeezed.unsqueeze(dim=0)
tensor_C_Squeezed, tensor_C_Unsqueezed, tensor_C_Unsqueezed.shape
```

```python
tensor_C_Unsqueezed = tensor_C_Squeezed.unsqueeze(dim=1)
tensor_C_Squeezed, tensor_C_Unsqueezed, tensor_C_Unsqueezed.shape
```

```python
x_original = torch.rand(size=(224, 224, 3))
x_permuted = x_original.permute(2, 0, 1)
print(f"Original shape: {x_original.shape}")
print(f"Permuted shape: {x_permuted.shape}")
```
